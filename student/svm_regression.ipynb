# Student score prediction using support vector regression (SVR)

I will use a linear kernel model with 10 folds, to compare against the mlp example

```python
import pandas as pd
from sklearn.svm import SVR
from sklearn.pipeline import Pipeline
from sklearn.model_selection import KFold, cross_val_score
from numpy import mean

import preprocess
import importlib

importlib.reload(preprocess)

from preprocess import preprocess

def process_train_eval(df):
    #Pre-process data
    X, y, preproc = preprocess(df, False)

    svr_linear = SVR(kernel='linear', C=100, epsilon=0.1)
    model = Pipeline([("prep",preproc),("mlp",svr_linear)])

    #Perform 10-fold cross validation with negative MSE as the scoring metric
    kfold = KFold(n_splits=10, shuffle=True, random_state=42)
    scores = cross_val_score(model, X, y, cv=kfold, scoring="neg_mean_squared_error")

    print(scores)
    print(scores.std())
    print(mean(scores))

df_maths = pd.read_csv("student-mat.csv", sep=";")

df_portuguese = pd.read_csv("student-por.csv", sep=";")

print("maths")
process_train_eval(df_maths)

print("portuguese")
process_train_eval(df_portuguese)
```
