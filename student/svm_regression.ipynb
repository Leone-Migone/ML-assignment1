```python
import pandas as pd
from preprocess import preprocess
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error
from sklearn.pipeline import Pipeline

def process_train_eval(df):
    X_train, X_test, y_train, y_test, preprocessor = preprocess(df)

    svr_linear = SVR(kernel='linear', C=100, epsilon=0.1)
    model_linear = Pipeline([("prep",preprocessor),("linear",svr_linear)])
    model_linear.fit(X_train, y_train)

    y_pred_linear = model_linear.predict(X_test)

    svr_rbf = SVR(kernel='rbf', C=100, gamma=0.5, epsilon=0.1)
    model_rbf = Pipeline([("prep",preprocessor),("rbf",svr_rbf)])

    model_rbf.fit(X_train,y_train)

    y_pred_rbf = model_rbf.predict(X_test)

    print("MSE linear:")
    print(mean_squared_error(y_test, y_pred_linear))

    print("MSE RBF:")
    print(mean_squared_error(y_test, y_pred_rbf))


df_maths = pd.read_csv("student-mat.csv", sep=";")

df_portuguese = pd.read_csv("student-por.csv", sep=";")

print("maths")
process_train_eval(df_maths)

print("portuguese")
process_train_eval(df_portuguese)
```
