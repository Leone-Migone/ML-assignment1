# Random Forest

This notebook covers analysis of the student dataset using a Random Forest to predict final grades.

First (G1) and second (G2) term grades are removed from the feature set, due to high correlation with the final grade.

Evaluation metrics: Mean MSE, Standard Deviation

```python
import pandas as pd
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
import preprocess
import importlib

importlib.reload(preprocess)

from preprocess import preprocess
```

```python
def process_train_eval(df):
    # Get the full dataset and preprocessor
    X, y, preprocessor = preprocess(df, split_test_train=False)

    # Create the Pipeline combining preprocessing and the Random Forest model
    pipeline = Pipeline(steps=[
         ('preprocessor', preprocessor),
         ('model', RandomForestRegressor(n_estimators=100, random_state=42))
     ])

    # Set up 10-Fold Cross-Validation
    cv = KFold(n_splits=10, shuffle=True, random_state=42)

    # Evaluate the model using mean squared error
    neg_mse_scores = cross_val_score(pipeline, X, y, cv=cv, scoring='neg_mean_squared_error')
    mse_scores = -neg_mse_scores

    print(f"MSE Scores per fold: {mse_scores}")
    print(f"Mean MSE: {mse_scores.mean():.4f}")
    print(f"Standard Deviation of MSE: {mse_scores.std():.4f}\n")

df_maths = pd.read_csv("student-mat.csv", sep=";")

df_portuguese = pd.read_csv("student-por.csv", sep=";")

print("Maths:")
process_train_eval(df_maths)

print("Portuguese:")
process_train_eval(df_portuguese)
```
