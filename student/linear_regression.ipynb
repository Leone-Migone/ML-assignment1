```python
import pandas as pd
from sklearn.model_selection import cross_val_predict, KFold
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import r2_score, mean_squared_error


def fit_subject(path):
    df = pd.read_csv(path, sep=";")
    # Target variable is final grade
    y = df["G3"]
    X = df.drop(columns=list(["G1", "G2", "G3"]))
    cat = X.select_dtypes(exclude="number").columns.to_list()
    pre = ColumnTransformer([("c", OneHotEncoder(drop="first"), cat)],
                            remainder="passthrough")
    model = make_pipeline(pre, LinearRegression())
    pred = cross_val_predict(model, X, y, cv=KFold(10, shuffle=True, random_state=0))
    print(f"R2={r2_score(y, pred):.3f}  MSE={mean_squared_error(y, pred):.2f}")
    return model.fit(X, y)

mat_model = fit_subject("student-mat.csv")
por_model = fit_subject("student-por.csv")
```
