```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df = pd.read_csv('agaricus-lepiota.data')
# use one hot encoding to convert categorical variables into numerical variables
# drop veil type column because it has only one value
df = df.drop(columns=['veil-type'])
print(f"Rows: {len(df):,} | Original columns: {df.shape[1]}")
df.head()
#use one hot encoding to convert categorical variables into numerical variables
numeric_df = pd.get_dummies(df, columns=df.columns.drop('poisonous'), dtype = int)
numeric_df['poisonous'] = numeric_df['poisonous'].map({'p': 1, 'e': 0})

assert numeric_df.select_dtypes(exclude="number").empty
assert not numeric_df.isna().any().any()
print(f"Shape: {numeric_df.shape} (all numeric)")
display(numeric_df.head())
#export it to csv
numeric_df.to_csv('./agaricus-lepiota-numeric.csv', index=False)







```
