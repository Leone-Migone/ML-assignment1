```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt


df = pd.read_csv('agaricus-lepiota-numeric.csv')

#split the dataset into features and the target variable class
df.head()
X = df.drop('poisonous',axis=1)
Y = df['poisonous']
#initialize kfold cross validation and classifier
kf = KFold(n_splits=5, shuffle=True, random_state=33)
clf_entropy = DecisionTreeClassifier(criterion="entropy", random_state=33, max_depth=5, min_samples_leaf=5)

#Perform cross validation and calculate prediction scores
y_pred = cross_val_predict(clf_entropy, X, Y, cv=kf)

# print confusion matrix
print(confusion_matrix(Y, y_pred))
# print classification report
class_report = classification_report(Y, y_pred)
print("Classification Report:\n", class_report)
```
