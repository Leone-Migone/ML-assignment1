```python
from sklearn.neural_network import MLPClassifier
import pandas as pd
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.metrics import classification_report, confusion_matrix 
df = pd.read_csv('agaricus-lepiota-numeric.csv') 
df.head()
#features
X=df.drop('poisonous', axis=1) 
#target 
Y = df['poisonous']
model = MLPClassifier(hidden_layer_sizes=(10, 10), max_iter=1000, random_state=33)

#initialize kfold cross validation and classifier
kf = KFold(n_splits=5, shuffle=True, random_state=33)
# use kfold and get predictions for each fold
y_pred = cross_val_predict(model, X, Y, cv=kf)

# print confusion matrix
print(confusion_matrix(Y, y_pred))
# print classification report
class_report = classification_report(Y, y_pred)
print("Classification Report:\n", class_report)
    
```
