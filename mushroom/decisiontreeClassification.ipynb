```python
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import KFold, cross_val_score
import matplotlib.pyplot as plt

df = pd.read_csv('agaricus-lepiota-numeric.csv')

#split the dataset into features and the target variable class
df.head()
X = df.drop('poisonous',axis=1)
Y = df['poisonous']
#initialize kfold cross validation and classifier
kf = KFold(n_splits=5, shuffle=True, random_state=42)
clf_entropy = DecisionTreeClassifier(criterion="entropy", random_state=33, max_depth=3, min_samples_leaf=5)

for fold, (train_index, test_index) in enumerate(kf.split(X)):
    #split the data into training and testing sets for this fold
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    Y_train, Y_test = Y.iloc[train_index], Y.iloc[test_index]
    #fit the classifier on the training data
    clf_entropy.fit(X_train, Y_train)
    #predict the target variable on the testing data
    y_pred = clf_entropy.predict(X_test)
    score = accuracy_score(Y_test, y_pred)
    print(f"Fold {fold + 1} Accuracy: {score:.4f}")

scores = cross_val_score(clf_entropy, X, Y, cv=kf, scoring='accuracy')

#Print individual scores, the average, and the variance
print(f"Scores per fold: {scores}")
print(f"Average Accuracy: {scores.mean():.4f} (+/- {scores.std() * 2:.4f})")








```
